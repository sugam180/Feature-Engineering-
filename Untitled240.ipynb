```python
import pandas as pd

from sklearn.preprocessing import OrdinalEncoder
from sklearn.preprocessing import LabelEncoder
```

```python
df.head()
```

```python
df['Membership_Level'].unique()
```

```python
df['Purchased'].unique()
```

```python
ordinal = OrdinalEncoder(
    categories=[['Bronze', 'Silver', 'Gold', 'Platinum']]
)
```

```python
df['Membership_Level'] = ordinal.fit_transform(
    df[['Membership_Level']]
)
```

```python
df.head()
```

```python
label = LabelEncoder()

df['Purchased'] = label.fit_transform(
    df['Purchased']
)
```
