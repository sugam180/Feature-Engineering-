```python
import pandas as pd
import numpy as np
import seaborn as sns
```

```python
df=pd.read_csv('/content/skewed_iqr_dataset (1).csv')
```

```python
sns.histplot(
    df["Purchase_Amount"],
    kde=True
)
```

```python
sns.boxplot(
    y=df["Purchase_Amount"]
)
```

```python
Q1 = df["Purchase_Amount"].quantile(0.25)

median = df["Purchase_Amount"].quantile(0.50)

Q3 = df["Purchase_Amount"].quantile(0.75)

print("Q1:", Q1)
print("Median:", median)
print("Q3:", Q3)
```

```python
IQR = Q3 - Q1

print("IQR:", IQR)
```

```python
lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)
```

```python
outliers = df[
    (df["Purchase_Amount"] < lower_limit) |
    (df["Purchase_Amount"] > upper_limit)
]

print(outliers)
```

```python
df["Purchase_Amount"] = np.where(
    df["Purchase_Amount"] > upper_limit,
    upper_limit,
    np.where(
        df["Purchase_Amount"] < lower_limit,
        lower_limit,
        df["Purchase_Amount"]
    )
)

print(df)
```
